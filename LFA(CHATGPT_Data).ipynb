```python
import pandas as pd

url = "https://dataset.raid-bench.xyz/train.csv"

chatgpt_chunks = []

for chunk in pd.read_csv(url, chunksize=100000):

    # Keep only ChatGPT rows
    chatgpt_chunk = chunk[
        chunk["model"].str.contains("chatgpt", case=False, na=False)
    ]

    chatgpt_chunks.append(chatgpt_chunk)

chatgpt_df = pd.concat(chatgpt_chunks, ignore_index=True)

print("Shape:", chatgpt_df.shape)
print(chatgpt_df["model"].value_counts())


# Dataset summary
print(f"Loaded Shape: {chatgpt_df.shape}")

```
